# Web of Science Starter API (CLARIVATE)


Laut des [offiziellen Dokuemnt zur Web of Science Starter API](https://developer.clarivate.com/apis/wos-starter) kann man folgerndermassen die Abfrage senden:

5 Abfragen in 1 Sekunde / 5'000 Abfragen pro Tag

Durch diese API-Schnittstelle erhalten wir sowohl die Zeitschriftendaten als auch die Daten der einzelnen Dokumenten erhalten.  

## Wie formuliert man die Abfrage?

- [Swagger](https://api.clarivate.com/swagger-ui/?apikey=none&url=https%3A%2F%2Fdeveloper.clarivate.com%2Fapis%2Fwos-starter%2Fswagger)
- [Query-Builder](https://webofscience.zendesk.com/hc/en-us/articles/20130361503249-Advanced-Search-Query-Builder#h_01JGCED24132Q8P7BBN7Z4YCMB)







````python
import requests
import getpass
import json
````

````python
CLARIVATE_KEY = getpass.getpass()
````

````python
WOS_DOCS_ENDPOINT = "https://api.clarivate.com/apis/wos-starter/v1/documents"
````

````python
wos_headers = {
    "X-ApiKey": CLARIVATE_KEY
}
````

## Beispiel-Query

Hier sind die Dokumente mit Suchkriterium "preferred organization name = University of Zurich" gesucht.  
Gefundene Dokumente werden nach Publikationsjahr absteigend sortiert. Dann werden 10 ersten Dokumente gezeigt.

````python
wos_query_params = {
    "q": "OG=(University of Zurich)",
    "sortField": "PY+D",
    "limit": 10
    }
````

````python
wos_resp = requests.get(WOS_DOCS_ENDPOINT, headers=wos_headers, params=wos_query_params)
wos_resp.status_code
````

````python
from IPython.display import JSON

JSON(wos_resp.json())
````

````python
wos_query_params2 = {
    "q": "OG=(University of Zurich) AND PY=2025 AND TS=mathematics",
    "sortField": "PY+D",
    "limit": 10
    }
````

````python
wos_resp2 = requests.get(WOS_DOCS_ENDPOINT, headers=wos_headers, params=wos_query_params2)
wos_resp2.status_code
````

````python
JSON(wos_resp2.json())
````
