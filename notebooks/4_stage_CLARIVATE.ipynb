```python
import requests
import getpass
import json
```

```python
CLARIVATE_KEY = getpass.getpass()
```

```python
WOS_ENDPOINT = "https://api.clarivate.com/apis/wos-starter/v1/documents"
```

```python
wos_headers = {
    "X-ApiKey": CLARIVATE_KEY
}
```

```python
wos_query_params = {
    "q": "OG=(University of Zurich)",
    "limit": 10
    }
```

```python
wos_resp = requests.get(WOS_ENDPOINT, headers=wos_headers, params=wos_query_params)
wos_resp.status_code
```

```python
from IPython.display import JSON

JSON(wos_resp.json())
```
